## 2. Un qubit en superposición

`H|0> = (|0> + |1>)/√2 = |+>`

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

qc = QuantumCircuit(1, 1)
qc.h(0)            # Hadamard: superposición
qc.measure(0, 0)   # medir
qc.draw('mpl')
```

## 3. Ejecutar y medir 1000 veces

```python
sim = AerSimulator()
conteos = sim.run(qc, shots=1000).result().get_counts()
print(conteos)
plot_histogram(conteos)
```

## 4. El estado en la esfera de Bloch

```python
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

qc2 = QuantumCircuit(1)
qc2.h(0)
plot_bloch_multivector(Statevector(qc2))
```
