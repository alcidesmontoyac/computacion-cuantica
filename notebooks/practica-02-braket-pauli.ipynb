## 0. Preparación

```python
import numpy as np
np.set_printoptions(precision=3, suppress=True)
```

## 1. Kets, bras y producto interno

Un qubit es un vector columna en \(\mathbb{C}^2\). El **bra** \(\langle\psi|\) es su conjugado transpuesto.

```python
# base computacional
ket0 = np.array([1, 0], dtype=complex)
ket1 = np.array([0, 1], dtype=complex)

# |psi> = (|0> + i|1>)/sqrt(2)
psi = (ket0 + 1j*ket1) / np.sqrt(2)
print("psi =", psi)

# <psi|psi> debe ser 1  (np.vdot conjuga el primer argumento)
print("<psi|psi> =", np.vdot(psi, psi).real)
```

## 2. Regla de Born

La probabilidad de medir el resultado asociado a \(|k\rangle\) es \(|\langle k|\psi\rangle|^2\).

```python
alpha = np.vdot(ket0, psi)   # <0|psi>
beta  = np.vdot(ket1, psi)   # <1|psi>
print("P(0) =", abs(alpha)**2)
print("P(1) =", abs(beta)**2)
print("suma =", abs(alpha)**2 + abs(beta)**2)
```

## 3. Matrices de Pauli

Las tres matrices de Pauli son los observables de espín. Todas son hermíticas con autovalores \(\pm1\).

```python
I = np.array([[1,0],[0,1]], dtype=complex)
X = np.array([[0,1],[1,0]], dtype=complex)
Y = np.array([[0,-1j],[1j,0]], dtype=complex)
Z = np.array([[1,0],[0,-1]], dtype=complex)

for nombre, M in [("X",X), ("Y",Y), ("Z",Z)]:
    valores, vectores = np.linalg.eigh(M)   # eigh: hermíticas
    print(nombre, "autovalores:", valores.real)
```

Los autovectores de \(Z\) son \(|0\rangle,|1\rangle\); los de \(X\) son \(|+\rangle,|-\rangle\). Comprobémoslo:

```python
valores, vectores = np.linalg.eigh(X)
print("autovectores de X (en columnas):")
print(vectores)
print("\n|+> esperado:", (ket0+ket1)/np.sqrt(2))
```

## 4. Medir en distintas bases

El mismo estado da resultados distintos según la base en que se mida.

```python
plus  = (ket0 + ket1)/np.sqrt(2)   # |+>
minus = (ket0 - ket1)/np.sqrt(2)   # |->

# |+> medido en la base Z {|0>,|1>}
print("|+> en base Z:  P(0)=", abs(np.vdot(ket0,plus))**2,
      " P(1)=", abs(np.vdot(ket1,plus))**2)

# |0> medido en la base X {|+>,|->}
print("|0> en base X:  P(+)=", abs(np.vdot(plus,ket0))**2,
      " P(-)=", abs(np.vdot(minus,ket0))**2)
```

## 5. No conmutatividad

Que \(X\) y \(Z\) no conmuten es la raíz del principio de incertidumbre.

```python
conmutador = X@Z - Z@X
print("[X,Z] =")
print(conmutador)
print("\n-2i*Y =")
print(-2j*Y)
```
